## Review

```python
def mult_three_neg(prompt: str, lower: float, upper: float) -> float | None:
    """
    Asks a user a question, gets input from the user, and returns the input if its a negative float divisble by three within the bounds.
    If they enter anything else is entered, None is returned

    Args:
        question (str): Question a user is prompted with
        lower (float): Lower bounds of acceptable answer
        upper (float): Upper bounds of acceptable answer

    Returns:
        (float | None) -> Float if user input is valid, else None
    """  # noqa: E501
    try:
        num = float(input(prompt))
    except ValueError:
        print('Non-valid input')
        return None
    if num < 0 and num % 3 == 0 and lower <= num <= upper:
        return num
    else:
        print('Non-valid input')
        return None


val = mult_three_neg('Enter a number: ', -20, -2)
if val:
    print(f'{val} is negative, bounded by -20 & -2 and was a multple of three')
```

## While for User Input Validation

```python
age = None
while True:
    inp = input('Please enter your age(positive integer > 0 )')
    try:
        iinp = int(inp)
        if iinp > 0:
            age = iinp
            break
        else:
            print('Age must be greater than 0')
            continue
    except ValueError:
        print('Age must be a valid integer')
        continue

print(f'Your age is {age}')
```

## Combining While with Functions

```python
def get_age(prompt: str) -> int:
    while True:
        inp = input(prompt)
        try:
            iinp = int(inp)
            if iinp > 0:
                return iinp
            else:
                print('Age must be greater than 0')
                continue
        except ValueError:
            print('Age must be a valid integer')
            continue


age = get_age('Please enter your age: ')
print(f'Your age is {age}')
```

## Extra Practice

1) Countdown
    - Story: print every whole number from `n` down to `0`.
    - Signature: `def countdown(n: int) -> None`

    - Story: given a `target` number and a list of `guesses` (simulating a player's guesses in order), return how many guesses it took to hit the target.
    - Signature: `def guess_number(target: int, guesses: list) -> int`

3) Average until 'done'
    - Story: given a list of number-strings ending in the sentinel `"done"`, average all the numbers that came before it.
    - Signature: `def average_until_done(entries: list) -> float`

    - Story: add up the individual digits of a whole number (e.g. `123` -> `1 + 2 + 3 = 6`) using a `while` loop, no string conversion.

### My Solutions

```python
def countdown(n: int) -> None:
    while n >= 0:
        print(n)
        n -= 1


def guess_number(target: int, guesses: list) -> int:
    attempts = 0
    gi = iter(guesses)
    while True:
        guess = next(gi)
        attempts += 1
        if guess == target:
            return attempts


def average_until_done(entries: list) -> float:
    total = 0.0
    count = 0
    ei = iter(entries)
    while True:
        entry = next(ei)
        if entry == 'done':
            break
        total += float(entry)
        count += 1
    return total / count if count else 0.0


def sum_of_digits(n: int) -> int:
    n = abs(n)
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total


def check_password(target: str, attempts_list: list, max_attempts: int = 3) -> bool:
    attempts = 0
    ai = iter(attempts_list)
    while attempts < max_attempts:
        attempt = next(ai)
        attempts += 1
        if attempt == target:
            return True
    return False


countdown(3)
print(guess_number(7, [3, 5, 7]))
print(average_until_done(['80', '90', '100', 'done']))
print(sum_of_digits(12345))
print(check_password('hunter2', ['wrong', 'hunter2']), check_password('hunter2', ['a', 'b', 'c']))
```
